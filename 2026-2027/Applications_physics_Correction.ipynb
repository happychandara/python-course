# **Exercise — Can We Verify Ohm’s Law with Python?**

A student wants to determine whether a resistor follows Ohm’s law. They measure the voltage U across the resistor for different current values I.

The experimental measurements are affected by small measurement errors.

According to Ohm’s law:

$$
U=R*I
$$

If the resistor follows this law, plotting voltage (U) against current (I) should give a straight line passing approximately through the origin. Its slope corresponds to the resistance (R).

***Part A — Experimental Data***

The student obtains the following measurements:

```python
I = np.array([0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90, 1.00])

U = np.array([1.1, 1.9, 3.2, 3.8, 5.2, 5.9, 7.1, 7.8, 9.2, 9.8])
```

1. Plot the experimental voltage (U) as a function of current (I), using a scatter plot.

2. Add a title, axis labels with units, and a grid.

3. Does the graph appear to be a straight line?

4. Do the measurements seem consistent with Ohm’s law? Explain your answer.

```python
plt.scatter(U,I)
plt.title("Experimental voltage in function of current")
plt.xlabel("U (V)")
plt.ylabel("I (A)")
plt.grid()
plt.show()
```

***Part B — Estimating the Resistance with Monte Carlo***

The following algorithm estimates the slope of the experimental data. It repeatedly selects random pairs of measurements, calculates the slope between them, and averages the results.

The algorithm is provided: you do not need to create it.

```python
def monte_carlo_slope(I, U, N=10000):
    slopes = []

    for _ in range(N):
        i, j = np.random.choice(len(I), 2, replace=False)

        if I[j] != I[i]:
            slope = (U[j] - U[i]) / (I[j] - I[i])
            slopes.append(slope)

    return np.mean(slopes), np.std(slopes)
```

1. Run the algorithm to estimate the slope and its standard deviation:

```python
R_est, uncertainty = monte_carlo_slope(I, U)

print("Estimated resistance:", R_est, "ohms")
print("Slope standard deviation:", uncertainty, "ohms")
```

2. Add the estimated straight line to your graph:

```python
I_line = np.linspace(0, 1.1, 100)
U_line = R_est * I_line

plt.scatter(I, U, label="Experimental measurements")
plt.plot(I_line, U_line, label="Monte Carlo slope")
plt.xlabel("Current I (A)")
plt.ylabel("Voltage U (V)")
plt.title("Experimental Verification of Ohm's Law")
plt.grid()
plt.legend()
plt.show()
```
