# **Exercise 3**: Fibonacci Sequence
The Fibonacci sequence is a sequence of numbers, defined as follows :
$$
F_{n+2}=F_{n+1}+F_{n} , \qquad with \qquad F_0=0 \qquad  and \qquad F_1=1
$$
Create an algorithm that displays the n first numbers of this sequence and test it with n=5 and n=10.

```python
F=[0,1]
n=int(input("Enter a number:"))
for k in range(2, n+1):
  F.append(F[k-1]+F[k-2])
print(F)
```

# **Exercise 7**: Recursive functions

A recursive function is a function that calls itself until it reaches a stopping condition, called the base case.
 - Always define a base case, otherwise the function loops forever
- Each call must get closer to the base case
- Python has a default recursion limit of 1000 calls
For example:


```python
def recursive_function(n):
    if n == base_case:      # Stop condition
        return some_value
    return recursive_function(n - 1)  # Self-call, moving toward base case
```

1. Build a factorial function that computes n! recursively. Try it with n=5.

```python
def factorial(n):
  if n==1:
    return 1
  else:
    return n*factorial(n-1)

print(factorial(5))
```

2. Build a Fibonacci function that returns the nth Fibonacci number recursively

```python
def Fibonacci(n):
  if n==0:
    return 0
  elif n==1:
    return 1
  else:
    return Fibonacci(n-1)+Fibonacci(n-2)
```

# **Exercise 9**: Calculate the average of a list
Create a function that calculates the average of a list of numbers. It should take a list as input, and return the average as output.

```python
def average(L):
  n=len(L)
  s=0
  for x in L:
    s=s+x
  return s/n

print(average([1,2,3]))
```

# **Exercise 10**: Sum of two matrices
Create 2 matrices :
- M , a matrix 3X3 that takes the numbers 1 to 9 included in ascending order
- N, a matrix 3X3 that takes the numbers 9 to 1 included in descending order

Create the matrix S, which is the addition of the matrices M and N : the coefficient with coordinates (1,1)
 of the matrix M is added to the coefficient with the same coordinates of the matrix N, and gives the coefficient (1,1)
 of the matrix S.

```python
M=[[1,2,3],[4,5,6],[7,8,9]]
N=[[9,8,7],[6,5,4],[3,2,1]]
S = [[0]*3 for _ in range(3)]
for i in range (len(M)):
  for j in range (len(M[0])):
    S[i][j]=M[i][j]+N[i][j]
print(S)
```

# **Exercise 11**: Count elements
Write a function count_elements that counts how many times each element appears in a list and returns a dictionary of the result, with the key being the element, and the value the number of times it appears. Test it with the list ["a", "b", "a", "c", "d", "a", "d", "a"].

```python
def count_elements(L):
  dico={}
  for x in L:
    if x in dico:
      dico[x]+=1
    else:
      dico[x]=1
  return dico
```

# **Exercise 12** : Concatenate dictionaries
Write a function merge_dicts that merges two dictionaries. If a key exists in both, add their values. Test it with {"a": 1, "b": 2} and {"b": 3, "c": 4}.

```python
def merge_dicts(dico1,dico2):
  fusion=dico1.copy()
  for x in dico2:
    if x in fusion:
      fusion[x]+=dico2[x]
    else:
      fusion[x]=dico2[x]
```

# **Exercise 14**: Merge two lists
Create a function taht merges two separate lists into a single, unified list. Test it with ["Banana", "Apple", "Peach"] and ["Mango", "Watermelon"]. It should return ["Banana", "Apple", Peach", "Mango", "Watermelon"].

```python
def merge(L1,L2):
  M=L1
  for x in L2:
    if x not in M:
      M.append(x)
  return M
```

# **Exercise 15**: List sorted
Create a function that determines if a list of numbers is sorted in non-decreasing (ascending) order. Return True if it is, and False otherwise. Test it with [1, 2, 6, 8, 9] and [2, 7, 4, 6].

```python
def sorted(L):
  for i in range (len(L)-1):
    if L[i]>L[i+1]:
      return False
  return True

print(sorted([1,2,6,8,9]))
print(sorted([2,7,4,6]))
```

# **Exercise 18**: Letter frequency
Write a function that returns a dictionary of the letter frequencies of a string. For instance, for the input hello, the function will return {"h": 1, "e": 1, "l": 2, "o": 1}. Test it with hello world.

```python
def letter_frequencies(s):
  freq={}
  for letter in s:
    if letter in freq:
      freq[letter]+=1
    else:
      freq[letter]=1
  return freq

print(letter_frequencies("hello world"))
```
