# **Exercise 3 — Matrix Calculations**

Two matrices represent measurements collected by two different sensors.

```python
A = np.array([[4, 2, 1],[3, 5, 2],[1, 0, 6]])

B = np.array([[2, 1, 3],[0, 4, 2],[5, 2, 1]])
```

1. Compute the element-wise difference between A and B.
2. Calculate the element-wise product of A and B.
3. Calculate the matrix product of A and B.
4. Compute the transpose of B.
5. Calculate the sum of the elements in each row of A.
6. Find the largest element in each column of B.

```python
def difference(A,B):
  M=np.zeros(3)
  for i in range(len(A)):
    for j in range(len(A[0])):
      M[i][j]=A[i][j]-B[i][j]
  return M

def product(A,B):
  M=np.zeros(3)
  for i in range(len(A)):
    for j in range(len(A[0])):
      M[i][j]=A[i][j]*B[i][j]
  return M

def matrix_product(A,B):
  return A@B

def transpose(A):
  return A.T

def sum(A):
  S=[]
  for i in range(len(A)):
    d=0
    for j in range len(A[i]):
      d=d+A[i][j]
    S.append(d)
  return S

def max_colums(A):
  max=[]
  for j in range(len(A[0])):
    m=A[0][j]
    for i in range(len(A)):
      if A[i][j]>m:
        m=A[i][j]
    max.append(m)
  return max

```

# **Exercise 4 — A Function for Matrix Multiplication (Difficult)**
Write a function called matrix_product(A, B) that computes the matrix product of two compatible two-dimensional NumPy arrays.

***Requirements:***

- The function must check whether the dimensions are compatible for matrix

multiplication.

- If the dimensions are incompatible, raise a ValueError.

- Otherwise, return the product of the two matrices.

- Test the function with matrices of shapes 2 × 3 and 3 × 2.

- Test the function with two incompatible matrices and verify that the error is raised.

- You may use loops and NumPy indexing, but do not use np.dot, np.matmul or the @ operator inside your function.

```python
def matrix_product(A, B):
    # Check whether the matrices are compatible
    if A.shape[1] != B.shape[0]:
        raise ValueError("Incompatible matrix dimensions")

    # Get the dimensions
    rows = A.shape[0]
    cols = B.shape[1]
    common = A.shape[1]

    # Initialize the result matrix
    result = np.zeros((rows, cols))

    # Calculate matrix multiplication using loops
    for i in range(rows):
        for j in range(cols):
            for k in range(common):
                result[i, j] += A[i, k] * B[k, j]

    return result
```
